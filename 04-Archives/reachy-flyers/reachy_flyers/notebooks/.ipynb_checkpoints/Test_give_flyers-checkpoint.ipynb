```python
import numpy as np 
import time

from reachy import Reachy, parts
from behavior.actions import Manipulate_flyer
```

```python
reachy = Reachy(
    head=parts.Head(io='/dev/ttyUSB*'),
    right_arm=parts.RightArm(io='/dev/ttyUSB*', hand='force_gripper'),
    left_arm=parts.LeftArm(io='/dev/ttyUSB*', hand='force_gripper')
)
```

```python
manip = Manipulate_flyer(reachy)
```

```python
def compliant(reachy):
    for m in reachy.right_arm.motors:
        m.compliant = True
    for m in reachy.head.motors:
        m.compliant = True
    for m in reachy.left_arm.motors:
        m.compliant = True
    reachy.head.compliant = True
```

```python
def stiff(reachy):
    for m in reachy.right_arm.motors:
        m.compliant = False
    for m in reachy.head.motors:
        m.compliant = False
    for m in reachy.left_arm.motors:
        m.compliant = False
    reachy.head.compliant = False
```

```python
stiff(reachy)
```

```python
manip.play('grab_flyer')
```

```python
compliant(reachy)
```

```python
for m in reachy.right_arm.motors:
    m.compliant = True 
```

```python
reachy.right_arm.hand.gripper.goal_position = 90
```
