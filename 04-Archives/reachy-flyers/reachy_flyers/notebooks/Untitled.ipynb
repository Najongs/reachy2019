```python
from reachy import parts, Reachy
import time
from pyquaternion import Quaternion
import numpy as np
```

```python
reachy = Reachy(
right_arm=parts.RightArm(io='/dev/ttyUSB*',hand='force_gripper'),
left_arm=parts.LeftArm(io='/dev/ttyUSB*',hand='flyer_hand'),
head=parts.Head(io='/dev/ttyUSB*')
)
```

```python
reachy.head.compliant = False
```

```python
q = reachy.head.neck.model.find_quaternion_transform([1, 0, 0], [0.5, 0, 0])
thetas = reachy.head.neck.model.get_angles_from_quaternion(q.w, q.x, q.y, q.z)
reachy.head.neck.disk_bottom.target_rot_position = thetas[2]
reachy.head.neck.disk_middle.target_rot_position = thetas[1]
reachy.head.neck.disk_top.target_rot_position = thetas[0]
```

```python
qA = Quaternion(axis=[1, 0, 0], angle=np.deg2rad(20))
reachy.head.neck.orient(qA, duration=1, wait=True)
```

```python
q1 = Quaternion(axis=[1, 0, 0], angle=np.deg2rad(20))
q2 = Quaternion(axis=[0, 1, 0], angle=np.deg2rad(25))
qB = q1*q2
reachy.head.neck.orient(qB, duration=1, wait=True)
```

```python
reachy.head.look_at(0.5,0,0, duration=1, wait=True)
```

```python
q = reachy.head.neck.model.find_quaternion_transform([1, 0, 0], [0.5, 0.2, 0])
qA = Quaternion(axis=[1, 0, 0], angle=np.deg2rad(15))
q = q*qA
```

```python
reachy.head.neck.orient(q, duration=1, wait=True)
```
